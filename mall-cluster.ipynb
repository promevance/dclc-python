# Univariate Analysis

```python
df.describe()
```

```python
# plotting distribution of Annual Income
sns.distplot(df['Annual Income (k$)'])

# distplot means distribution plot and it is used to plot the distribution of a variable.
# distplot is a combination of a histogram and a kernel density estimate (KDE) plot.
# The histogram shows the frequency of data points in different bins, while the KDE plot shows the estimated probability density function of the data.
# distplot is deprecated in recent versions of seaborn, and it is recommended to use histplot or displot instead.
```

```python
for i in df.columns:
    plt.figure()
    sns.histplot(df[i], kde=True)
    plt.title(f'Distribution of {i}')
    plt.show()
```

```python
# KDE plot for Annual Income. KDE stands for Kernel Density Estimate which is the line over the histogram

sns.kdeplot(df['Annual Income (k$)'], fill=True)
```

```python
# plotting kdeplot for annual income and comparing it to that for Gender
sns.kdeplot(data=df, x='Annual Income (k$)', hue='Gender', fill=True)   
```

```python
# getting kdeplot for all the columns in the dataframe (compared to Gender hue), while excluding the Gender column
for i in df.columns:
    if i != 'Gender':
        plt.figure()
        sns.kdeplot(data=df, x=i, hue='Gender', fill=True)
        plt.title(f'KDE Plot of {i} by Gender')
        plt.show()
```

```python
# Getting boxplots with Gender as x-axis for all columns except Gender
# male and female in Gender will be shown as two different colors in the boxplot
for i in df.columns:
    if i != 'Gender':
        plt.figure()
        sns.boxplot(data=df, x='Gender', y=i)
        plt.title(f'Boxplot of {i} by Gender')
        plt.show()
```

# Clustering - Univariate, Bivariate, Multivariate

```python
clustering1=KMeans(n_clusters=5, random_state=42)
clustering1.fit(df2[['Annual Income (k$)', 'Spending Score (1-100)']])
df2['Cluster_Income_Spending']=clustering1.labels_
df2.head()
```

```python
clustering1.inertia_
```

```python
# calculating inertia for different values of k to use the elbow method
inertia_values = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df2[['Annual Income (k$)', 'Spending Score (1-100)']])
    inertia_values.append(kmeans.inertia_)
```

```python
inertia_values
```

```python
# plotting the elbow method graph
plt.plot(range(1, 11), inertia_values, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.show()
```

```python
df2.groupby('Cluster_Income_Spending').mean(numeric_only=True)
```

```python
clustering2=KMeans(n_clusters=5, random_state=42)
clustering2.fit(df2[['Age', 'Spending Score (1-100)']])
df2['Cluster_Age_Spending']=clustering2.labels_
df2.head()
```

```python
inertia_values2 = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df2[['Age', 'Spending Score (1-100)']])
    inertia_values2.append(kmeans.inertia_)
inertia_values2
```

```python
# plotting the elbow method graph for Age vs Spending Score
plt.plot(range(1, 11), inertia_values2, marker='o')
plt.title('Elbow Method for Optimal k (Age vs Spending Score)')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.show()
```

```python
sns.scatterplot(data=df2, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster_Income_Spending', palette='Set1')
```

```python
# getting the centroids of the clusters
centroids = clustering1.cluster_centers_
centres=pd.DataFrame(centroids, columns=['Annual Income (k$)', 'Spending Score (1-100)'])
```

```python
centres
```

```python
# plotting clusters with centroids
plt.figure(figsize=(10,6))
sns.scatterplot(data=centres, x='Annual Income (k$)', y='Spending Score (1-100)', s=200, color='black', label='Centroids')
sns.scatterplot(data=df2, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster_Income_Spending', palette='Set1')
plt.title('Clusters of Annual Income vs Spending Score with Centroids')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.savefig('clusters_with_centroids.png')
plt.show()
```

```python
# cross tabulation between clusters and Gender
pd.crosstab(df2['Cluster_Income_Spending'], df2['Gender'],normalize='index')
```

```python
#multivariate clustering
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler = StandardScaler()
```

```python
df2.head()
```

```python
df3 = pd.get_dummies(df2, columns=['Gender'], drop_first=True)
df3.head()
```

```python
df3_columns = df3.columns.tolist()
df3_columns.remove('Cluster_Income_Spending')
df3_columns
```

```python
df3=df3[['Age','Annual Income (k$)','Spending Score (1-100)','Gender_Male']]
df3.head()
```

```python
df3=scaler.fit_transform(df3)
```

```python
df3=pd.DataFrame(scaler.fit_transform(df3))
df3.head()
```

```python
# calculating and plotting inertia for different values of k for multivariate clustering
inertia_values3 = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df3)
    inertia_values3.append(kmeans.inertia_)
plt.plot(range(1, 11), inertia_values3, marker='o')
plt.title('Elbow Method for Optimal k (Multivariate Clustering)')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.savefig('elbow_multivariate.png')
plt.show()
```
