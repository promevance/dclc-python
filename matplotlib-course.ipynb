```python
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
income = [15000, 17000, 18000, 26000, 29000, 31000, 33000, 42000, 44000, 45000]  

income_ticks = list(range(10000, 50001, 5000))  # Custom y-axis ticks from 10,000 to 50,000 with step of 5,000

plt.plot(years, income, color='green', linestyle='solid')
plt.title("Annual Income Over Years", fontsize=14, fontname='Arial')
plt.xlabel("Years")
plt.ylabel("Income (in USD)")
plt.yticks(income_ticks, [f"${i}" for i in income_ticks])  # Custom y-axis labels
```

```python
from matplotlib import style

style.use('ggplot') # Use ggplot style for better aesthetics
style.use('dark_background') # Use dark background style

votes = {'Alice': 120, 'Bob': 95, 'Charlie': 130, 'Diana': 110}
names = list(votes.keys())
values = list(votes.values())

plt.pie(values, autopct='%1.1f%%', startangle=90)
plt.legend(names, title="Candidates", loc="best")  # Add legend to the pie chart instead of labels on slices
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.  
plt.title("Vote Distribution Among Candidates")
plt.show()
```

```python
x = np.arange(100) # the arange function creates an array of values from 0 to 99

fig, axs = plt.subplots(2,2)  # Create a 2x2 grid of subplots

axs[0, 0].plot(x, np.sin(x), 'tab:blue')
axs[0, 0].set_title('Sine Wave')

axs[0, 1].plot(x, np.cos(x), 'tab:orange')
axs[0, 1].set_title('Cosine Wave')

axs[1, 0].plot(x, np.tan(x), 'tab:green')
axs[1, 0].set_title('Tangent Wave')

axs[1, 1].plot(x, np.log(x + 1), 'tab:red')  # Added +1 to avoid log(0)
axs[1, 1].set_title('Logarithmic Curve')

plt.tight_layout()  # Adjust subplots to fit into figure area.

# plt.show() # Display all the figures. only needed if not using an interactive environment

# to save a figure without plotting it to the screen
plt.savefig('four_plots.png', dpi=300, bbox_inches='tight', transparent=False) # transparent = True for transparent background
# dpi: resolution of the saved figure
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')


%matplotlib widget

x=np.arange(-5,5,0.1)
y=np.arange(-5,5,0.1)
X, Y = np.meshgrid(x, y) # create a grid of x and y values
Z = np.sin(X) * np.cos(Y)  # compute Z values based on X and Y grid

ax.plot_surface(X, Y, Z) # cmap='plasma', edgecolor='none')
ax.set_title('3D Surface Plot')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.show()
```

```python
import plotly.graph_objects as go
import numpy as np

x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y)])
fig.update_layout(title="Interactive 3D Surface", autosize=True)
fig.show()
```

```python
import random
import matplotlib.pyplot as plt

heads_tails = [0, 0]

for _ in range(10000):
    heads_tails[random.randint(0, 1)] += 1

plt.bar(['Heads', 'Tails'], heads_tails, color=['blue', 'orange'], edgecolor='black')
plt.title("Heads vs Tails (10,000 flips)")
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.show()
plt.pause(0.001) # animation effect
```

```python
# Recreating the above animation using Plotly

import random
import plotly.graph_objects as go

# Number of flips
num_flips = 200

# Keep running totals
heads = 0
tails = 0

# Lists to store animation frames
frames = []
x = ["Heads", "Tails"]

for i in range(1, num_flips + 1):
    # Simulate flip
    if random.randint(0, 1) == 0:
        heads += 1
    else:
        tails += 1
    
    # Add frame for this flip count
    frames.append(
        go.Frame(
            data=[go.Bar(x=x, y=[heads, tails])],
            name=str(i)
        )
    )

# Create initial figure
fig = go.Figure(
    data=[go.Bar(x=x, y=[0, 0])],
    layout=go.Layout(
        title="Coin Flip Simulation: Growing Counts Over Time",
        xaxis=dict(title="Outcome"),
        yaxis=dict(title="Count", range=[0, num_flips]),
        updatemenus=[
            dict(
                type="buttons",
                showactive=True,
                buttons=[
                    dict(
                        label="Play",
                        method="animate",
                        args=[None, {"frame": {"duration": 30, "redraw": True},
                                     "fromcurrent": True}]
                    ),
                    dict(
                        label="Pause",
                        method="animate",
                        args=[[None], {"frame": {"duration": 0}, "mode": "immediate"}]
                    )
                ]
            )
        ]
    ),
    frames=frames
)

fig.show()

    
```
